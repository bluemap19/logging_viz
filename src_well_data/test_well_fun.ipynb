```python
T0 = time.time()
well = DATA_WELL(path_folder = r'Z:\logging_workspace\姬119H2')
```

```python
# =====================================================================
# 1. 文件扫描与总览
# =====================================================================
print("\n" + "=" * 70)
print("【1】well_summary() — 井数据总览")
print("=" * 70)
summary = well.well_summary()
for k, v in summary.items():
    print(f"  {k}: {v}")
```

```python
# =====================================================================
# 2. get_path_list_* 系列 — 各类型文件路径列表
# =====================================================================
print("\n" + "=" * 70)
print("【2】get_path_list_* 系列 — 文件路径列表")
print("=" * 70)
print("logging files:", len(well.get_path_list_logging()), "个")
for p in well.get_path_list_logging(): print("  -", p)
print("table files:", len(well.get_path_list_table()), "个")
for p in well.get_path_list_table(): print("  -", p)
print("fmi files:", len(well.get_path_list_fmi()), "个")
for p in well.get_path_list_fmi(): print("  -", p)
print("nmr files:", len(well.get_path_list_nmr()), "个")
for p in well.get_path_list_nmr(): print("  -", p)
print("core files:", len(well.get_path_list_core()), "个")
for p in well.get_path_list_core(): print("  -", p)
```

```python
# =====================================================================
# 3. search_*_path_list 系列 — 关键字精确搜索
# =====================================================================
print("\n" + "=" * 70)
print("【3】search_*_path_list 系列 — 关键字精确搜索")
print("=" * 70)
p_logging = well.search_file_path_list(name_keywords=['姬119H2', '常规测井'])
p_table = well.search_file_path_list(name_keywords=['table'])
p_dyna = well.search_file_path_list(name_keywords=['DYNA'])
p_stat = well.search_file_path_list(name_keywords=['STAT'])
p_nmr = well.search_file_path_list(name_keywords=['NMR'])
p_core = well.search_file_path_list(name_keywords=['CORE'])

print("search_logging_path_list(['姬119H2','常规测井']):", p_logging)
print("search_table_path_list(['table']):", p_table)
print("search_fmi_path_list(['DYNA']):", p_dyna)
print("search_fmi_path_list(['STAT']):", p_stat)
print("search_nmr_path_list(['NMR']):", p_nmr)
print("search_core_path_list(['CORE']):", p_core)
```

```python
# =====================================================================
# 4. init_* 系列 — 惰性初始化（触发日志输出）
# =====================================================================
print("\n" + "=" * 70)
print("【4】init_* 系列 — 惰性初始化")
print("=" * 70)
try:
    well.init_logging(p_logging[0] if p_logging else ''); print("  init_logging() OK")
except Exception as e: print(f"  init_logging() FAIL: {e}")
try:
    well.init_table(p_table[0] if p_table else ''); print("  init_table() OK")
except Exception as e: print(f"  init_table() FAIL: {e}")
try:
    if p_dyna: well.init_FMI(p_dyna[0]); print("  init_FMI() OK")
    else: print("  init_FMI() SKIP (no DYNA file)")
except Exception as e: print(f"  init_FMI() FAIL: {e}")
try:
    if p_nmr: well.init_NMR(p_nmr[0]); print("  init_NMR() OK")
    else: print("  init_NMR() SKIP (no NMR file)")
except Exception as e: print(f"  init_NMR() FAIL: {e}")
try:
    if p_core: well.init_core(p_core[0]); print("  init_core() OK")
    else: print("  init_core() SKIP (no CORE file)")
except Exception as e: print(f"  init_core() FAIL: {e}")
```

```python
# =====================================================================
# 5. get_logging — 常规测井数据获取
# =====================================================================
print("\n" + "=" * 70)
print("【5】get_logging — 常规测井数据")
print("=" * 70)
try:
    df_log = well.get_logging(
        key=p_logging[0] if p_logging else '',
        curve_names=['#DEPTH', 'GR', 'DEN', 'CNL', 'AC', 'SP'],
        norm=False,
        depth_limit=[2726, 2760],
    )
    print(f"  Shape: {df_log.shape}")
    print(f"  Columns: {list(df_log.columns)}")
    print("  Head:\n" + df_log.head(3).to_string())
    print("  Describe:\n" + df_log.describe().to_string())
except Exception as e:
    print(f"  FAIL: {e}")
```

```python
# =====================================================================
# 6. get_table — 岩性类型表（mode='2' 通常更稳定）
# =====================================================================
print("\n" + "=" * 70)
print("【6】get_table — 岩性类型表")
print("=" * 70)
try:
    df_tab2 = well.get_table(key=p_table[0])
    print(f"  mode='2' Shape: {df_tab2.shape}")
    print("  Head:\n" + df_tab2.head(5).to_string())
except Exception as e:
    print(f"  mode='2' FAIL: {e}")
try:
    df_tab3 = well.get_table(key=p_table[1])
    print(f"  mode='3' Shape: {df_tab3.shape}")
    print("  Head:\n" + df_tab3.head(5).to_string())
except Exception as e:
    print(f"  mode='3' FAIL: {e}")

try:
    rep_dict = well.get_table_replace_dict(table_key=p_table[0])
    print("  get_table_replace_dict():", rep_dict)
except Exception as e:
    print(f"  get_table_replace_dict() FAIL: {e}")
```

```python
# =====================================================================
# 7. get_core — 岩心实验数据
# =====================================================================
print("\n" + "=" * 70)
print("【7】get_core — 岩心实验数据")
print("=" * 70)
try:
    df_core = well.get_core(
        key=p_core[0] if p_core else '',
        curve_names=['DEPTH', '石英', '钾长石', '黄铁矿'],
        depth_range=[2726, 2760],
    )
    print(f"  Shape: {df_core.shape}")
    print(f"  Columns: {list(df_core.columns)}")
    print("  Data:\n" + df_core.to_string())
except Exception as e:
    print(f"  FAIL: {e}")
```

```python
# =====================================================================
# 8. get_NMR — 核磁共振数据
# =====================================================================
print("\n" + "=" * 70)
print("【8】get_NMR — 核磁共振数据")
print("=" * 70)
if p_nmr:
    try:
        nmr_result = well.get_NMR(key=p_nmr[0], depth=[2726, 2760])
        print(f"  Type: {type(nmr_result)}")
        if nmr_result is not None and isinstance(nmr_result, tuple):
            print(f"  NMR image shape: {nmr_result[0].shape}")
    except Exception as e:
        print(f"  FAIL: {e}")
else:
    print("  [跳过] 未扫描到 NMR 文件")
```

```python
# =====================================================================
# 9. get_FMI — 电成像数据
# =====================================================================
print("\n" + "=" * 70)
print("【9】get_FMI — 电成像数据")
print("=" * 70)
if p_dyna:
    try:
        fmi_result = well.get_FMI(key=p_dyna[0], depth=[2726, 2760])
        print(f"  Type: {type(fmi_result)}")
        if fmi_result is not None and isinstance(fmi_result, tuple):
            print(f"  FMI image shape: {fmi_result[0].shape}")
            print(f"  FMI depth shape: {fmi_result[1].shape if len(fmi_result) > 1 else 'N/A'}")
    except Exception as e:
        print(f"  FAIL: {e}")
else:
    print("  [跳过] 未扫描到 FMI DYNA 文件")
```

```python
# =====================================================================
# 10. get_FMI_fde — 电成像 FDE 数据
# =====================================================================
print("\n" + "=" * 70)
print("【10】get_FMI_fde / get_FMI_textures — 电成像 FDE/纹理（计算慢，可选）")
print("=" * 70)
tex_cfg = {'window': 200, 'step': 100}
print(f"    texture_config = {tex_cfg}")
if p_dyna:
    try:
        fde_result = well.get_FMI_fde(key=p_dyna[0], fde_config=tex_cfg)
        print(f"  Type: {type(fde_result)}")
        if fde_result is not None and isinstance(fde_result, tuple):
            print(f"  FMI image shape: {fde_result[0].shape}")
            print(f"  FMI depth shape: {fde_result[1].shape if len(fde_result) > 1 else 'N/A'}")
    except Exception as e:
        print(f"  FAIL: {e}")
else:
    print("  [跳过] 未扫描到 FMI DYNA 文件， 无法进行FDE计算")
# 快速验证: 调用一次 get_FMI_fde（只取第一帧做样本，0.5s）
```

```python
# =====================================================================
# 11. combine_logging_table — logging + table 合并
# =====================================================================
print("\n" + "=" * 70)
print("【11】combine_logging_table — logging + table 合并")
print("=" * 70)
try:
    df_merged_tab = well.combine_logging_table(
        logging_key=p_logging[0] if p_logging else '',
        curve_names_logging=['#DEPTH', 'GR', 'DEN', 'CNL'],
        table_key=p_table[0] if p_table else '',
        new_col='Type',
    )
    print(f"  Shape: {df_merged_tab.shape}")
    print(f"  Columns: {list(df_merged_tab.columns)}")
    print("  Head:\n" + df_merged_tab.head(5).to_string())
except Exception as e:
    print(f"  FAIL: {e}")
```

```python
# =====================================================================
# 12. combine_logging_tables — logging + core 合并
# =====================================================================
print("\n" + "=" * 70)
print("【12】combine_logging_tables — logging + core 合并")
print("=" * 70)
try:
    df_merged_core = well.combine_logging_core(
        logging_key=p_logging[0] if p_logging else '',
        curve_names_logging=['#DEPTH', 'GR', 'DEN', 'CNL'],
        core_key=p_core[0] if p_core else '',
        depth_limit=[2726, 2760],
        tolerance=0.5,
    )
    print(f"  Shape: {df_merged_core.shape}")
    print(f"  Columns: {list(df_merged_core.columns)}")
    matched = df_merged_core.dropna(subset=['石英'])
    print(f"  有矿物数据的行: {len(matched)} / {len(df_merged_core)}")
    print("  矿物数据示例:")
    print(matched[['#DEPTH', 'GR', '石英', '钾长石', '黄铁矿']].head(5).to_string(index=False))
except Exception as e:
    print(f"  FAIL: {e}")
```

```python
# =====================================================================
# 13. combine_logging_tables — 仅 logging（退化场景）
# =====================================================================
print("\n" + "=" * 70)
print("【13】combine_logging_tables — 仅 logging（退化场景）")
print("=" * 70)
try:
    df_only_log = well.combine_logging_table_core(
        logging_key=p_logging[0] if p_logging else '',
        curve_names_logging=['#DEPTH', 'GR', 'DEN'],
        depth_limit=[2730, 2745],
        table_key=p_table[0],
        core_key=p_core[0],
    )
    print(f"  Shape: {df_only_log.shape}")
    print(df_only_log.head(3).to_string())
except Exception as e:
    print(f"  FAIL: {e}")

print("\n" + "=" * 70)
print(f"全部测试完成，耗时 {time.time() - T0:.2f}s")
print("=" * 70)
```
